```python
import numpy as np
```

```python
#sigmoid activation function
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
#step 2 inputs
x1=0.6
x2=0.8
```

```python
#step 3 hidden layer weights and bias
w1=0.4
w2=0.7
b1=0.2
```

```python
#step 4: Hidden Neuron calculation
z1=(x1*w1)+(x2*w2)+b1
h1=sigmoid(z1)
```

```python
#Output layer weights and bias
w3=0.9
b2=0.1
```

```python
#step 6 Output Neuron calculation
z2=(h1*w3)+b2
y1=sigmoid(z2)
```

```python
#step 7 :Actual output
t1=1
```

```python
#step 8 : Error and Loss calculation
error=t1-y1
loss=0.5*(error**2)
```

```python
#Display the results
print("==== Forward Propagation Results ===")
print("Hidden Net Input (z1):", z1)
print("Hidden Neuron Output (h1):", h1)
print("Output Net Input (z2):", z2)
print("Output Neuron Output (y1):", y1)

print("Actual Output (t1):", t1)
print("Error:", error)
print("Loss:", loss)
```
