```python
import numpy as np
```

```python
#sigmoid function
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
#sigmoid derivative function
def sigmoid_derivative(x):
    return x*(1-x)
```

```python
#inputs
x1=0.6
x2=0.8
```

```python
w1=0.4
w2=0.7
b1=0.2
```

```python
w3=0.9
b2=0.1
```

```python
#target
t1=1
```

```python
#learning rate
lr=0.1
```

```python
#trainging loop
for i in range(50):
    ##forward pass
    z1=x1*w1+x2*w2+b1
    h1=sigmoid(z1)
    z2=h1*w3+b2
    y1=sigmoid(z2)
```

```python
#Error
error=t1-y1
```

```python
#backpropagation
delta_y1=error*sigmoid_derivative(y1)
delta_h1=delta_y1*w3*sigmoid_derivative(h1)
```

```python
#Update weights and biases
w1=w1+lr*delta_h1*x1
w2=w2+lr*delta_h1*x2
b1=b1+lr*delta_h1
```

```python
if i%10==0:
    print(f"Iteration {i}: Error: {error}, Output: {y1}")
    
```

```python
print(f"Final weights: w1={w1}, w2={w2}, b1={b1}, w3={w3}, b2={b2}")
print(f"Final output: {y1}, Final error: {error}")
#print("Loss ",loss)
```
